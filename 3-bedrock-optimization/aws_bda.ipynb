# Amazon Bedrock Data Automation (BDA)

## Introdução
O Amazon Bedrock Data Automation (BDA) simplifica a extração de insights estruturados de uma ampla gama de tipos de conteúdo não estruturado, incluindo documentos, imagens, vídeos e áudio.

## Modos de Saída
- **Extração Padrão**: 
  - Baseada nos padrões do Bedrock.
  - Disponível para todos os tipos de conteúdo.
  - Configurações configuráveis por tipo de conteúdo.

- **Saída Personalizada**: 
  - Regras de extração definidas pelo usuário.
  - Disponível apenas para documentos e imagens.
  - Utiliza blueprints para lógica de extração.

## Componentes Principais
### Projetos
- **Definição**: Coleção de configurações.
- **Características**:
  - Opções de saída padrão configuráveis.
  - Suporte a múltiplos blueprints.
  - Capacidades de divisão de documentos.
  - Configurações de otimização de recursos.

### Blueprints
- **Propósito**: Instruções para extração de dados personalizada.
- **Tipos**:
  - Blueprints de exemplo (pré-construídos).
  - Blueprints personalizados (criados pelo usuário).
- **Tipos de Inferência**:
  - **Explícita**: Extração direta de documentos.
  - **Implícita**: Inferência de valor contextual.

## Casos de Uso
- Indexação RAG (Retrieval Augmented Generation).
- Processamento de documentos.
- Análise de mídia.
- Classificação de conteúdo.
- Moderação de conteúdo.

## Implementação Técnica
### API
- **Método**: `invoke_data_automation_async`
- **Parâmetros Requeridos**:
  - Localização de entrada S3.
  - Localização de saída S3.
  - Configuração do projeto.
  - Blueprint(s).

### Recursos Avançados
- **Processamento de Documentos**:
  - Divisão inteligente de documentos.
  - Detecção de limites semânticos.
  - Correspondência de blueprints com base no layout.

- **Análise de Conteúdo**:
  - Reconhecimento de texto em imagens.
  - Detecção de logotipos.
  - Resumo de cenas.

## Exemplo Prático
### Caso de Uso: Processamento de Certificado de Certificação AWS
1. Criar blueprint personalizado.
2. Fazer upload do documento de exemplo.
3. Definir regras de extração.
4. Configurar parâmetros de saída.
5. Processar documentos usando o blueprint.

# Exercício: Análise de Vídeo com Amazon Bedrock

## Visão Geral do Exercício
Neste exercício, você irá:
- Capturar e enviar um vídeo curto de si mesmo discutindo qualquer tópico.
- Usar a automação de dados do Amazon Bedrock para analisar seu vídeo, gerando resumos, transcrições, detectando logotipos e etiquetando conteúdo com metadados como categorias IAB.
- Explorar e refinar os resultados da análise, aprendendo como o Bedrock pode impulsionar a inteligência de mídia e ser integrado em fluxos de trabalho de processamento de conteúdo escaláveis e do mundo real.

## Tarefas

### Tarefa 1: Filmando seu Vídeo
- **Duração**: Tente filmar cerca de 30 segundos de conteúdo.
- **Tema**: Você pode falar sobre qualquer coisa, como seu filme favorito, o que comeu no café da manhã ou por que seu animal de estimação é o melhor.
- **Logotipo da Marca**: Se possível, segure um logotipo de marca durante o vídeo.

### Tarefa 2: Enviando seu Vídeo
1. Abra o console da AWS e faça login usando suas credenciais de `exerciseuser`.
2. No console de gerenciamento da AWS, busque e escolha o Amazon Bedrock.
3. No painel de navegação, sob Automação de Dados, selecione **Ver Demonstração**.
4. Sob **Começar**, escolha **Enviar do Computador** e selecione seu vídeo filmado recentemente.
5. Escolha **Gerar resultado**. Se for a primeira vez que você usa a automação de dados com o Bedrock, uma caixa de diálogo de confirmação aparecerá. Selecione **Confirmar** para criar um bucket S3 para armazenar o vídeo.

### Tarefa 3: Analisando o Conteúdo do Seu Vídeo
#### Tarefa 3.1: Primeira Análise
- Revise o que o Bedrock descobriu sobre seu vídeo. O resumo gerado foi preciso?
- Sob Configuração, selecione todas as caixas possíveis e escolha **Regenerar resultados**. Dê o nome **im-a-star** ao seu projeto e escolha **Criar Projeto BDA**.

#### Tarefa 3.2: Segunda Análise
- O que o Bedrock encontrou desta vez? A transcrição está correta? As etiquetas IAB são precisas? O Bedrock identificou logotipos que você incluiu no vídeo?

### Tarefa 4: Limpeza
1. No console do Bedrock, selecione **Projetos** e escolha **im-a-star**, depois selecione **Excluir**.
2. No console da AWS, busque e escolha **S3**. Selecione o bucket que se parece com: `bedrock-bda-us-east-1-c0adb7fb-cc6d-480c-81c7-5a7fbb6e5598`.
3. Escolha **Esvaziar** e digite **deletar permanentemente**. Selecione **Esvaziar**.
4. Exclua o bucket novamente, digitando o nome do bucket e selecionando **Excluir Bucket**.

## Resumo
Você agora sabe como:
- Enviar e processar mídia usando a automação de dados do Amazon Bedrock.
- Verificar e melhorar os resultados, aprendendo como o Bedrock pode ajudar a entender vídeos automaticamente e economizar tempo.
